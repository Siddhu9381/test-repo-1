# Notebook 1: Absolute Imports

This notebook demonstrates **absolute imports** from the utils and models packages.

## Import Statements

We'll import functions, classes, and methods using absolute import paths.

```python
# Absolute imports - importing specific functions
from utils.math_utils import add, subtract, multiply, divide
from utils.string_utils import capitalize_words, reverse_string, count_words
```

```python
# Absolute imports - importing classes
from utils.data_processor import DataProcessor
from models.base_model import BaseModel
from models.user_model import UserModel
```

```python
# Absolute imports - importing entire modules
import utils.math_utils as math_utils
import utils.string_utils as string_utils
```

## Using Imported Functions

```python
# Using directly imported functions
result1 = add(10, 5)
result2 = subtract(20, 8)
result3 = multiply(6, 7)
result4 = divide(100, 4)

print(f"Add: {result1}")
print(f"Subtract: {result2}")
print(f"Multiply: {result3}")
print(f"Divide: {result4}")
```

```python
# Using module-level imports
result5 = math_utils.add(15, 25)
result6 = string_utils.capitalize_words("hello world from notebook")

print(f"Module import add: {result5}")
print(f"Module import capitalize: {result6}")
```

## Using Imported Classes and Methods

```python
# Using imported DataProcessor class
processor = DataProcessor("notebook data")
print(f"Processed: {processor.process()}")
print(f"Transformed: {processor.transform('notebook-transform')}")
print(f"Valid: {processor.validate()}")
```

```python
# Using imported UserModel class
user = UserModel("Notebook User", "notebook@example.com", "nb-001")
print(f"User info: {user.get_full_info()}")
print(f"User dict: {user.to_dict()}")
print(f"Save result: {user.save()}")
```
