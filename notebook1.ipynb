## Defining Functions in Notebooks

We can define our own functions in notebooks that use imported modules.

```python
# Define a function that uses imported functions
def process_text_data(text):
    """
    Process text data using imported string utilities.
    
    This function demonstrates using imported functions within a notebook function.
    
    Args:
        text: Input text string
    
    Returns:
        Dictionary with processed text information
    """
    # Use imported string utility functions
    capitalized = capitalize_words(text)
    reversed_text = reverse_string(text)
    word_count = count_words(text)
    
    return {
        'original': text,
        'capitalized': capitalized,
        'reversed': reversed_text,
        'word_count': word_count
    }

# Test the function
result = process_text_data("hello world from notebook")
print("Text Processing Results:")
for key, value in result.items():
    print(f"  {key}: {value}")
```

```python
# Define a function that performs calculations using imported math functions
def calculate_statistics(numbers):
    """
    Calculate basic statistics for a list of numbers.
    
    This function demonstrates using multiple imported functions together.
    
    Args:
        numbers: List of numbers
    
    Returns:
        Dictionary with sum, average, and product
    """
    if not numbers:
        return {'sum': 0, 'average': 0, 'product': 0}
    
    # Calculate sum using imported add function
    total_sum = 0
    for num in numbers:
        total_sum = add(total_sum, num)
    
    # Calculate average
    count = len(numbers)
    average = divide(total_sum, count)
    
    # Calculate product using imported multiply function
    product = 1
    for num in numbers:
        product = multiply(product, num)
    
    return {
        'sum': total_sum,
        'average': average,
        'product': product,
        'count': count
    }

# Test the function
stats = calculate_statistics([2, 4, 6, 8, 10])
print("Statistics Results:")
for key, value in stats.items():
    print(f"  {key}: {value}")
```
