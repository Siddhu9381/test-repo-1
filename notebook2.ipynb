# Notebook 2: Relative and Mixed Imports

This notebook demonstrates **relative imports** and **mixed import patterns**.

Note: Relative imports in Jupyter notebooks require special handling. This notebook shows how to work with imports in notebook context.

## Setup for Relative Imports

In Jupyter notebooks, we need to add the parent directory to the path for relative imports to work properly.

```python
import sys
import os

# Add parent directory to path for imports
sys.path.insert(0, os.path.abspath('..'))
sys.path.insert(0, os.path.abspath('.'))
```

## Absolute Imports (Standard Pattern)

```python
# Absolute imports - most common in notebooks
from utils.math_utils import add, multiply
from utils.string_utils import capitalize_words
from models.user_model import UserModel
```

## Import Entire Modules

```python
# Import entire modules
import utils.math_utils
import utils.string_utils
import models.base_model
```

## Using Imported Functions

```python
# Direct function calls
sum_result = add(25, 17)
product_result = multiply(8, 9)
text_result = capitalize_words("notebook two demonstration")

print(f"Sum: {sum_result}")
print(f"Product: {product_result}")
print(f"Capitalized: {text_result}")
```

```python
# Module-level function calls
sum_module = utils.math_utils.add(50, 30)
text_module = utils.string_utils.reverse_string("notebook")

print(f"Module sum: {sum_module}")
print(f"Module reverse: {text_module}")
```

## Using Imported Classes

```python
# Create UserModel instance
user2 = UserModel("Alice Johnson", "alice@example.com", "nb2-001")

# Use methods from the class
print(f"Full info: {user2.get_full_info()}")
print(f"Dictionary: {user2.to_dict()}")
print(f"Save: {user2.save()}")
print(f"Delete: {user2.delete()}")
```

```python
# Using BaseModel from module import
base2 = models.base_model.BaseModel("nb2-base-123")
print(f"Base model: {base2.to_dict()}")
```

## Package-Level Imports

```python
# Import packages (uses __init__.py exports)
import utils
import models

# Use package-level exports
pkg_add = utils.add(100, 200)
pkg_capitalize = utils.capitalize_words("package level import")

print(f"Package add: {pkg_add}")
print(f"Package capitalize: {pkg_capitalize}")

# Create model from package
pkg_user = models.UserModel("Package User", "pkg@example.com")
print(f"Package user: {pkg_user.get_full_info()}")
```

## Import All from Module

```python
# Import all functions from a module
from utils.math_utils import *

# Now we can use functions directly without prefix
result_all = add(multiply(5, 6), divide(30, 2))
print(f"Complex calculation: {result_all}")
```

## Mixed Import Patterns

```python
# Mix of different import styles
from utils import DataProcessor  # From package __init__
from models.user_model import UserModel  # Direct from module
import utils.string_utils as str_utils  # Module alias

# Use mixed imports
proc = DataProcessor("mixed import data")
user_mixed = UserModel("Mixed User", "mixed@example.com")
text_mixed = str_utils.count_words("this is a mixed import test")

print(f"Processor: {proc.process()}")
print(f"User: {user_mixed.get_full_info()}")
print(f"Word count: {text_mixed}")
```
